```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
url = "https://archive.ics.uci.edu/static/public/186/data.csv"

columns = ['fixed_acidity','volatile_acidity','citric_acid','residual_sugar','chlorides','free_sulfur_dioxide','total_sulfur_dioxide','density','pH','sulphates','alcohol','quality','color']
df = pd.read_csv(url, names=columns)
df.head()
```

```python
print('Shape:', df.shape)
print('\nColumn Names:', df.columns.tolist())
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
before = df.shape[0]
df.drop_duplicates(inplace=True)
after = df.shape[0]
print('Duplicates removed:', before - after)
```

```python
encoder = LabelEncoder()
df['color'] = encoder.fit_transform(df['color'])

df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

cols = [
    'fixed_acidity','volatile_acidity','citric_acid','residual_sugar',
    'chlorides','free_sulfur_dioxide','total_sulfur_dioxide','density',
    'pH','sulphates','alcohol'
]

df[cols] = scaler.fit_transform(df[cols])

df.head()
```

```python
df.to_csv('assignment2_cleaned_iris.csv', index=False)
```
