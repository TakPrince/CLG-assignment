```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix


import pandas as pd
import numpy as np
```

```python

data = [
    [1, 'Male', 20, 78, 72, 75, 'Pass'],
    [2, 'Female', 19, 85, 80, 88, 'Pass'],
    [3, 'Male', 21, 60, 58, 62, 'Pass'],
    [4, 'Female', 20, 45, 50, 48, 'Fail'],
    [5, 'Male', 22, 30, 35, 40, 'Fail'],
    [6, 'Female', 21, 90, 92, 89, 'Pass'],
    [7, 'Male', 19, 70, 68, 72, 'Pass'],
    [8, 'Female', 20, 55, 60, 58, 'Fail'],
    [9, 'Male', 23, 88, 85, 90, 'Pass'],
    [10, 'Female', 22, 76, 74, 78, 'Pass'],
    [11, 'Male', 21, 66, 64, 68, 'Pass'],
    [12, 'Female', 20, 40, 45, 42, 'Fail'],
    [13, 'Male', 24, 95, 93, 96, 'Pass'],
    [14, 'Female', 19, 82, 79, 85, 'Pass'],
    [15, 'Male', 22, 58, 55, 60, 'Fail'],
    [16, 'Female', 21, 91, 89, 92, 'Pass'],
    [17, 'Male', 20, 73, 70, 75, 'Pass'],
    [18, 'Female', 23, 48, 52, 50, 'Fail'],
    [19, 'Male', 21, 67, 65, 69, 'Pass'],
    [20, 'Female', 22, 86, 83, 88, 'Pass']
]

columns = [
    'student_id',
    'gender',
    'age',
    'math_score',
    'science_score',
    'english_score',
    'result'
]

df = pd.DataFrame(data, columns=columns)

df.head()
```

```python
df['gender'] = df['gender'].map({'Male': 0, 'Female': 1})

# Features and target
X = df[['age', 'gender', 'math_score', 'science_score', 'english_score']]
y = df['result']

# Check shapes
print(X.shape, y.shape)
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Print sizes
print('Train size:', X_train.shape)
print('Test size:', X_test.shape)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
print('Model trained.')
```

```python
y_pred = knn.predict(X_test)
list(y_pred)[:10]
```

```python
results = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

results['Status'] = results.apply(lambda r: 'Correct' if r.Actual == r.Predicted else 'Wrong', axis=1)
results.head(20)
```

```python
print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nClassification Report:')
print(classification_report(y_test, y_pred))
print('\nConfusion Matrix:')
print(confusion_matrix(y_test, y_pred))
```
